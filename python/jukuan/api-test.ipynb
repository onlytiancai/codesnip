## 指数行情

```python
import pandas as pd

def get_zz_quote(code, end_date=None, count=None, start_date=None):
    '''获取中证指数行情
    返回 DataFrame,MultiIndex 为 (TradingDay, SecuCode)
    '''
    if isinstance(code, str):
        code = [code]

    code.sort()
    code = [x[:6] for x in code]

    days = get_trade_days(start_date, end_date, count)

    code_df = jy.run_query(query(
        jy.SecuMain.InnerCode, jy.SecuMain.SecuCode, jy.SecuMain.ChiName
    ).filter(
        jy.SecuMain.SecuCode.in_(code)
    ).order_by(jy.SecuMain.SecuCode))

    df = jy.run_query(query(
        jy.QT_CSIIndexQuote
    ).filter(
        jy.QT_CSIIndexQuote.IndexCode.in_(code_df.InnerCode),
        jy.QT_CSIIndexQuote.TradingDay.in_(days),
    ))

    df2 = pd.merge(code_df, df, left_on='InnerCode', right_on='IndexCode')
    df2 = df2.set_index(['TradingDay', 'SecuCode'])
    df2 = df2.drop(['InnerCode', 'IndexCode', 'ID', 'UpdateTime', 'JSID',
                    'OpenInterest', 'SettleValue', 'IndexCSIType'], axis=1)

    return df2  # 不再 .to_panel()
```

```python
df = get_zz_quote(['000016.XSHG','000001.XSHG'],end_date='2019-01-21',count=10)
```
